```python
import hashlib
```

```python
def sha256_hash(message):
    return hashlib.sha256(message.encode()).hexdigest()

original = input("Enter original message: ")
received = input("Enter received message: ")
```

```python
hash_original = sha256_hash(original)
hash_received = sha256_hash(received)

print("\nOriginal Hash:", hash_original)
print("Received Hash:", hash_received)
```

```python
if hash_original == hash_received:
    print("\nIntegrity Verified: Message is unchanged.")
else:
    print("\nIntegrity Failed: Message has been Alterd.")
```
